```python
import pandas as pd

# Load the dataset
df = pd.read_csv("data/titanic.csv")

# Display the first 5 rows
df.head()
```

```python
# Number of rows and columns
print("Dataset Shape:", df.shape)

# Column names
print("\nColumns:")
print(df.columns)

# Information about the dataset
print("\nDataset Information:")
df.info()
```

```python
# Check missing values
missing_values = df.isnull().sum()

print(missing_values)
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Survived", hue="Survived", palette="Set2", legend=False)

plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.savefig("images/survival_count.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

- The number of passengers who did not survive (0) is higher than those who survived (1).
- This indicates that the overall survival rate on the Titanic was less than 50%.
- The dataset is slightly imbalanced, with more non-survivors than survivors.

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="Sex", hue="Survived", palette="Set1")

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.savefig("images/survival_by_gender.png",dpi=300, bbox_inches="tight")
plt.show()
```

## Passenger Class Distribution

```python
plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Pclass",
    palette="viridis",
    legend=False
)

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.savefig("images/passenger_class_distribution.png",dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

- The majority of passengers traveled in **Third Class (Pclass = 3)**.
- **First Class** had the fewest passengers.
- This indicates that most passengers on the Titanic belonged to the third passenger class.

## Survival by Passenger Class

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived",
    palette="Set2"
)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.savefig("images/survival_by_class.png",dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

- First-class passengers had the highest number of survivors.
- Third-class passengers had the highest number of non-survivors.
- This suggests that passengers traveling in higher classes had a better chance of survival than those in lower classes.

## Age Distribution

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=30,
    kde=True,
    color="skyblue"
)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.savefig("images/age_distribution.png",dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

- Most passengers were between **20 and 40 years** old.
- There were relatively few elderly passengers (above 60 years).
- The dataset also includes infants and young children.
- The age distribution is slightly right-skewed, with fewer passengers in the higher age groups.

```python
## Fare Distribution
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Fare",
    bins=40,
    kde=True,
    color="orange"
)

plt.title("Fare Distribution of Titanic Passengers")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.savefig("images/fare_distribution.png",dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

- Most passengers paid relatively low fares, with a large concentration below 50.
- The fare distribution is highly right-skewed, meaning a small number of passengers paid very high fares.
- A few passengers paid exceptionally high fares, indicating the presence of outliers.
- Higher fares were generally associated with premium passenger classes.

## Correlation Heatmap

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select only numerical columns
numeric_df = df.select_dtypes(include=["number"])

# Compute correlation matrix
correlation_matrix = numeric_df.corr()

# Plot the heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")
plt.savefig("images/correlation_heatmap.png",dpi=300, bbox_inches="tight")
plt.show()
```

### Observation

- Passenger Class (`Pclass`) is negatively correlated with Fare, indicating that passengers in higher classes generally paid higher ticket fares.
- Survival has a positive correlation with Fare, suggesting passengers with higher fares had better survival chances.
- No pair of numerical features shows an extremely strong correlation, indicating that each feature provides different information.

# Final Conclusions

## Key Insights

1. **Overall Survival Rate**
   - Less than half of the passengers survived the Titanic disaster.
   - The dataset shows a higher number of non-survivors than survivors.

2. **Impact of Gender on Survival**
   - Female passengers had a significantly higher survival rate than male passengers.
   - Male passengers accounted for the majority of the casualties.

3. **Impact of Passenger Class**
   - First-class passengers had a higher chance of survival.
   - Third-class passengers experienced the highest number of fatalities, indicating that passenger class influenced survival.

4. **Passenger Distribution**
   - Most passengers traveled in Third Class, while First Class had the fewest passengers.

5. **Age Distribution**
   - The majority of passengers were between 20 and 40 years old.
   - The dataset also included children and elderly passengers, though they represented a smaller proportion of travelers.

6. **Fare Distribution**
   - Most passengers paid relatively low ticket fares.
   - The fare distribution was highly right-skewed, with a few passengers paying exceptionally high fares.

7. **Missing Values**
   - The **Age** column contained 177 missing values.
   - The **Cabin** column contained 687 missing values, making it the least complete feature.
   - The **Embarked** column contained only 2 missing values.

8. **Feature Relationships**
   - Passenger Class and Fare were negatively correlated, indicating that passengers in higher classes generally paid higher fares.
   - Survival showed a positive relationship with Fare and a negative relationship with Passenger Class, suggesting that socioeconomic status influenced survival chances.

## Overall Summary

The exploratory data analysis revealed that **gender**, **passenger class**, and **ticket fare** were the most influential factors associated with survival on the Titanic. Female passengers and those traveling in First Class had considerably higher survival rates than male passengers and those in Third Class. The dataset also contained missing values, particularly in the **Cabin** column, which should be addressed before developing predictive machine learning models.

This analysis provides a clear understanding of the dataset and forms a strong foundation for future data preprocessing, feature engineering, and predictive modeling.
